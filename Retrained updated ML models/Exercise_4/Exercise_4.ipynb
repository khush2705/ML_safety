```python
# ============================================================
# EXERCISE 4.5 - CELL 12
# OBSERVABLE ODD K-PROJECTION COVERAGE
# ============================================================

from kprojection import KProjectionCoverage

print("=" * 70)
print("EXERCISE 4.5 - OBSERVABLE ODD COVERAGE")
print("=" * 70)


# ------------------------------------------------------------
# Observable ODD dimensions
# ------------------------------------------------------------
#
# These are the dimensions for which our test dataset provides
# evidence:
#
#   Weather          -> dry
#   Lighting         -> daytime
#   Camera condition -> operational
#
# Scene type and vehicle speed are intentionally NOT included
# here because the test metadata does not provide them.
# ------------------------------------------------------------

observable_odd = {

    "weather": [
        "dry"
    ],

    "lighting": [
        "daytime"
    ],

    "camera_condition": [
        "operational"
    ]
}


# ------------------------------------------------------------
# Observed scenario
# ------------------------------------------------------------

observed_scenario = {
    "weather": "dry",
    "lighting": "daytime",
    "camera_condition": "operational"
}


print("\nObservable ODD:")
for dimension, values in observable_odd.items():
    print(f"  {dimension}: {values}")


print("\nObserved scenario:")
print(observed_scenario)


# ------------------------------------------------------------
# Calculate k = 1, 2, 3
# ------------------------------------------------------------

results = []

for k in [1, 2, 3]:

    coverage_metric = KProjectionCoverage(
        k=k,
        desc=observable_odd
    )

    coverage_metric.add_scenario(observed_scenario)

    result = coverage_metric.compute()

    results.append(result)

    print("\n" + "-" * 70)
    print(f"k = {k}")
    print("-" * 70)

    print("Coverage      :", result.coverage)
    print("Coverage (%)  :", result.coverage * 100)
    print("Covered points:", result.covered)
    print("Total points  :", result.total)
    print("Scenes        :", result.scenes)


print("\n" + "=" * 70)
print("OBSERVABLE ODD COVERAGE COMPLETE")
print("=" * 70)
```

```python
# ============================================================
# EXERCISE 4.5 - CELL 13
# FULL ODD PROJECTION SPACE
# ============================================================

from itertools import combinations
from math import prod

print("=" * 70)
print("FULL ODD PROJECTION SPACE")
print("=" * 70)

# This is the ODD defined from Exercise 2.
full_odd = {
    "weather": ["dry"],
    "lighting": ["daytime"],
    "camera_condition": ["operational"],
    "scene_type": [
        "urban_road",
        "mapped_intersection"
    ],
    "vehicle_speed": [
        "at_or_below_50_kmh"
    ]
}

dimensions = list(full_odd.keys())

print("\nODD dimensions:")
for dimension in dimensions:
    print(f"  {dimension}: {full_odd[dimension]}")


print("\n" + "-" * 70)
print("PROJECTION SPACE")
print("-" * 70)

for k in [1, 2, 3]:

    total_points = 0

    print(f"\nk = {k}")

    for projection_dims in combinations(dimensions, k):

        n_points = prod(
            len(full_odd[d])
            for d in projection_dims
        )

        total_points += n_points

        print(
            f"  {projection_dims} "
            f"-> {n_points} possible points"
        )

    print("  TOTAL projection points:", total_points)


print("\n" + "=" * 70)
print("FULL ODD PROJECTION SPACE COMPLETE")
print("=" * 70)
```

```python
# ============================================================
# EXERCISE 4.7 - CELL 15
# LOAD SAVED MODELS AND TEST DATA
# ============================================================

import os
import torch
import torch.nn as nn
import torchvision.models as models
from torchvision import transforms
from torch.utils.data import Dataset, DataLoader
from PIL import Image
import pandas as pd
import numpy as np

print("=" * 70)
print("EXERCISE 4.7 - MODEL AND TEST DATA LOADING")
print("=" * 70)


# ------------------------------------------------------------
# Device
# ------------------------------------------------------------

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("\nDevice:", device)

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))


# ------------------------------------------------------------
# Paths
# ------------------------------------------------------------

BASE_PATH = r"E:\ML_Safety"

TEST_DIR = os.path.join(
    BASE_PATH,
    "test"
)

MODEL_DIR = os.path.join(
    BASE_PATH,
    "model_outputs"
)

print("\nTest directory:")
print(TEST_DIR)

print("\nModel directory:")
print(MODEL_DIR)


# ------------------------------------------------------------
# Transformation
# ------------------------------------------------------------

test_transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])


# ------------------------------------------------------------
# Load labels
# ------------------------------------------------------------

labels_path = os.path.join(
    TEST_DIR,
    "labels.feather"
)

labels_df = pd.read_feather(labels_path)

print("\nLabel columns:")
print(labels_df.columns.tolist())

print("\nNumber of test labels:", len(labels_df))


# ------------------------------------------------------------
# Image lookup
# ------------------------------------------------------------

image_dir = os.path.join(
    TEST_DIR,
    "rgb-front"
)

image_files = sorted(
    [
        f for f in os.listdir(image_dir)
        if f.lower().endswith((".jpg", ".jpeg", ".png"))
    ]
)

print("\nNumber of test images:", len(image_files))


# ------------------------------------------------------------
# Create image lookup by frame ID
# ------------------------------------------------------------

image_lookup = {}

for filename in image_files:

    stem = os.path.splitext(filename)[0]

    try:
        frame_id = int(stem)
        image_lookup[frame_id] = os.path.join(
            image_dir,
            filename
        )

    except ValueError:
        pass


print("Image lookup entries:", len(image_lookup))


# ------------------------------------------------------------
# Dataset
# ------------------------------------------------------------

class SafetyTestDataset(Dataset):

    def __init__(
        self,
        dataframe,
        target_column,
        image_lookup,
        transform=None
    ):

        self.df = dataframe.reset_index(drop=True)
        self.target_column = target_column
        self.image_lookup = image_lookup
        self.transform = transform

    def __len__(self):
        return len(self.df)

    def __getitem__(self, idx):

        row = self.df.iloc[idx]

        frame_id = int(row["frame"])

        image_path = self.image_lookup[frame_id]

        image = Image.open(image_path).convert("RGB")

        if self.transform is not None:
            image = self.transform(image)

        label = int(row[self.target_column])

        return image, label


# ------------------------------------------------------------
# Model loader
# ------------------------------------------------------------

def create_model(model_path):

    model = models.resnet18(weights=None)

    model.fc = nn.Linear(
        model.fc.in_features,
        2
    )

    checkpoint = torch.load(
        model_path,
        map_location=device
    )

    if isinstance(checkpoint, dict) and "model_state_dict" in checkpoint:

        model.load_state_dict(
            checkpoint["model_state_dict"]
        )

    else:

        model.load_state_dict(checkpoint)

    model = model.to(device)
    model.eval()

    return model


# ------------------------------------------------------------
# Model paths
# ------------------------------------------------------------

model_paths = {

    "Pedestrian": os.path.join(
        MODEL_DIR,
        "best_pedestrian.pth"
    ),

    "Traffic Light": os.path.join(
        MODEL_DIR,
        "best_traffic_light.pth"
    ),

    "Vehicle": os.path.join(
        MODEL_DIR,
        "best_vehicle.pth"
    )
}


# ------------------------------------------------------------
# Load models
# ------------------------------------------------------------

loaded_models = {}

for name, path in model_paths.items():

    print(f"\nLoading {name}...")

    if not os.path.exists(path):
        raise FileNotFoundError(
            f"Model not found: {path}"
        )

    loaded_models[name] = create_model(path)

    print("Loaded successfully.")
    print("Device:", next(
        loaded_models[name].parameters()
    ).device)


print("\n" + "=" * 70)
print("ALL MODELS LOADED SUCCESSFULLY")
print("=" * 70)
```

```python
# ============================================================
# EXERCISE 4.7 - CELL 16
# GPU TEST-SET EVALUATION
# ============================================================

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

print("=" * 70)
print("EXERCISE 4.7 - PER-CLASS TEST EVALUATION")
print("=" * 70)


# ------------------------------------------------------------
# Configuration
# ------------------------------------------------------------

TARGET_COLUMNS = {

    "Pedestrian":
        "has_pedestrian",

    "Traffic Light":
        "has_traffic_light",

    "Vehicle":
        "has_vehicle"
}


BATCH_SIZE = 64
NUM_WORKERS = 0


# ------------------------------------------------------------
# Evaluation function
# ------------------------------------------------------------

def evaluate_model(
    model,
    dataframe,
    target_column,
    model_name
):

    dataset = SafetyTestDataset(
        dataframe=dataframe,
        target_column=target_column,
        image_lookup=image_lookup,
        transform=test_transform
    )

    loader = DataLoader(
        dataset,
        batch_size=BATCH_SIZE,
        shuffle=False,
        num_workers=NUM_WORKERS,
        pin_memory=torch.cuda.is_available()
    )


    all_labels = []
    all_predictions = []


    model.eval()


    with torch.no_grad():

        for images, labels in loader:

            # Move batch to GPU
            images = images.to(
                device,
                non_blocking=True
            )

            labels = labels.to(
                device,
                non_blocking=True
            )


            # Forward pass
            outputs = model(images)


            # Binary prediction
            predictions = torch.argmax(
                outputs,
                dim=1
            )


            # Move results back to CPU
            all_labels.extend(
                labels.cpu().numpy()
            )

            all_predictions.extend(
                predictions.cpu().numpy()
            )


    # Convert to NumPy
    y_true = np.array(all_labels)
    y_pred = np.array(all_predictions)


    # --------------------------------------------------------
    # Metrics
    # --------------------------------------------------------

    accuracy = accuracy_score(
        y_true,
        y_pred
    )

    precision = precision_score(
        y_true,
        y_pred,
        zero_division=0
    )

    recall = recall_score(
        y_true,
        y_pred,
        zero_division=0
    )

    f1 = f1_score(
        y_true,
        y_pred,
        zero_division=0
    )

    cm = confusion_matrix(
        y_true,
        y_pred,
        labels=[0, 1]
    )


    # --------------------------------------------------------
    # Print results
    # --------------------------------------------------------

    print("\n" + "=" * 60)
    print(f"{model_name.upper()} TEST RESULTS")
    print("=" * 60)

    print(f"Accuracy  : {accuracy:.4f}")
    print(f"Precision : {precision:.4f}")
    print(f"Recall    : {recall:.4f}")
    print(f"F1 Score  : {f1:.4f}")

    print("\nConfusion Matrix:")
    print(cm)


    return {
        "model": model_name,
        "accuracy": accuracy,
        "precision": precision,
        "recall": recall,
        "f1": f1,
        "confusion_matrix": cm,
        "y_true": y_true,
        "y_pred": y_pred
    }


# ------------------------------------------------------------
# Evaluate all three models
# ------------------------------------------------------------

evaluation_results = {}


for model_name, model in loaded_models.items():

    target_column = TARGET_COLUMNS[model_name]

    evaluation_results[model_name] = evaluate_model(
        model=model,
        dataframe=labels_df,
        target_column=target_column,
        model_name=model_name
    )


# ------------------------------------------------------------
# Summary table
# ------------------------------------------------------------

summary_rows = []

for model_name, result in evaluation_results.items():

    summary_rows.append({
        "Model": model_name,
        "Accuracy": result["accuracy"],
        "Precision": result["precision"],
        "Recall": result["recall"],
        "F1": result["f1"]
    })


evaluation_summary = pd.DataFrame(
    summary_rows
)


print("\n" + "=" * 70)
print("TEST SET PERFORMANCE SUMMARY")
print("=" * 70)

display(
    evaluation_summary.style.format({
        "Accuracy": "{:.4f}",
        "Precision": "{:.4f}",
        "Recall": "{:.4f}",
        "F1": "{:.4f}"
    })
)


print("\n" + "=" * 70)
print("EXERCISE 4.7 EVALUATION COMPLETE")
print("=" * 70)
```

```python
# ============================================================
# EXERCISE 4.7 - CELL 17
# CONFUSION MATRIX PLOTS
# ============================================================

import matplotlib.pyplot as plt

print("=" * 70)
print("EXERCISE 4.7 - CONFUSION MATRIX VISUALIZATION")
print("=" * 70)


for model_name, result in evaluation_results.items():

    cm = result["confusion_matrix"]

    plt.figure(figsize=(6, 5))

    plt.imshow(cm)

    plt.title(
        f"{model_name} - Confusion Matrix"
    )

    plt.xlabel("Predicted Label")
    plt.ylabel("True Label")

    plt.xticks(
        [0, 1],
        ["Negative", "Positive"]
    )

    plt.yticks(
        [0, 1],
        ["Negative", "Positive"]
    )

    # Add numerical values
    for i in range(2):
        for j in range(2):

            plt.text(
                j,
                i,
                str(cm[i, j]),
                ha="center",
                va="center"
            )

    plt.colorbar(
        label="Number of Samples"
    )

    plt.tight_layout()
    plt.show()


print("\n" + "=" * 70)
print("ALL THREE CONFUSION MATRICES PLOTTED")
print("=" * 70)
```
